```python
class matrix:
    def __init__(self, *args):
        if len(args) == 2 and isinstance(args[0], int) and isinstance(args[1], int):
            n, m = args[0], args[1]
            self.rows = n
            self.cols = m
            self.data = [[0 for _ in range(m)] for _ in range(n)]
        elif len(args) == 1 and isinstance(args[0], list):
            data = args[0]
            self.rows = len(data)
            self.cols= len(data[0])
            for row in data:
                if not isinstance(row, list):
                    raise TypeError("Error")
                if len(row) != self.cols:
                    raise TypeError("Error")
            self.data = data
    def __getitem__(self, key):
        if isinstance(key, tuple):
            i, j = key
            return self.data[i][j]
        elif isinstance(key, int):
            return self.data[key]
    def assign(self, other):
        if isinstance(other, matrix):
            if self.rows != other.rows or self.cols != other.cols:
                raise ValueError("Error")
            for i in range(self.rows):
                for j in range(self.cols):
                    self.data[i][j] = other.data[i][j]
        elif isinstance(other, list):
            if len(other)!= self.rows:
                raise ValueError("Row size error")
            for row in other:
                if not isinstance(row, list) or len(row) != self.cols:
                    raise ValueError("Column size error")
            for i in range(self.rows):
                for j in range(self.cols):
                    self.data[i][j] = other[i][j]
    def __str__(self):
        return str(self.data)
```

```python
def constant(n,m,c):
    return matrix([[float(c) for _ in range(m)] for _ in range(n)])
def zeros(n,m):
    return constant(n,m,0.0)
def ones(n,m):
    return constant(n,m,1.0)
def eye(n):
    matri = zeros(n,n)
    for i in range(n):
        matri.data[i][i] = 1.0
    return matri
```
